## Which country has produced the most content?

<h4>
split(", ") will make each row as a list. So it is useful if there is multiple value in a each row. Then I have used explode() to make each element in the list as a separate row.  

Here was another problem which is why I need to to split() explode() two times. 

After doing split() explode() I saw some row's value still has "," with it, which was making it a different row. For example: "United States" and "United States,". So I took the previously exploded df and split(",") it again. Thus the comma was eliminated and value_counts() gives content_count_per_country.

I have also kept the previous explode().value_counts() to see the difference.
</h4>

```python
country_df_split = df['country'].str.split(', ')
country_df_split.head(100)
```

```python
country_df_explode = country_df_split.explode('country')
country_df_explode
```

```python
content_per_country = country_df_explode.value_counts()
content_per_country
```

```python
country_df_split = country_df_explode.str.split(",")
country_df_split
```

```python
country_split_explode = country_df_split.explode('country')
country_split_explode
```

```python
content_count_per_country = country_split_explode.value_counts()
content_count_per_country
```
